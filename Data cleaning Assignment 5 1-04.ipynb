```python
data_subset = data.iloc[:, :58]
```

```python
data_subset.iloc[:, 57]= np.where(data_subset.iloc[:, 57] > 1, 1, data_subset.iloc[:, 57])
```

```python
data_subset
```

```python
numeric_data = data_subset.select_dtypes(include=[np.number])
```

```python
cols_with_all_na = numeric_data.columns[numeric_data.isna().all()]
numeric_data.drop(columns=cols_with_all_na, inplace=True)
```

```python
cols_with_na = numeric_data.columns[numeric_data.isna().any()]
for col in cols_with_na:
    mode_value = numeric_data[col].mode()[0]
    numeric_data[col].fillna(mode_value, inplace=True)
```
