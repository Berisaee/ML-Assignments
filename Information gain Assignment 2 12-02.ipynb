```python
yes_count = df['Class:Buys_computer'].value_counts().get('yes', 0)
no_count = df['Class:Buys_computer'].value_counts().get('no', 0)
total_entries = len(df) 

info_D = round(-(yes_count/total_entries)*math.log2(yes_count/total_entries)-(no_count/total_entries)*math.log2(no_count/total_entries),4)
print(info_D)
```

```python
info_D_values = {}
gain_values = {}

categorical_columns = df.columns[df.dtypes == 'object'].tolist()
categorical_columns.remove('Class:Buys_computer')

def calculate_info_D(column):
    unique_values = df[column].unique()
    info_D = 0

    for value in unique_values:
        subset = df[df[column] == value]
        count_value = len(subset)
        
        bought_computer_cnt = (subset['Class:Buys_computer'] == 'yes').sum()
        have_not_bought_computer_cnt = (subset['Class:Buys_computer'] == 'no').sum()
        
        p_yes = bought_computer_cnt / count_value if bought_computer_cnt > 0 else 1
        p_no = have_not_bought_computer_cnt / count_value if have_not_bought_computer_cnt > 0 else 1
        
        entropy = - (p_yes * math.log2(p_yes)) - (p_no * math.log2(p_no))
        info_D += (count_value / total_entries) * entropy
    
    return round(info_D, 4)

for col in categorical_columns:
    info_D_values[col] = calculate_info_D(col)
    gain_values[col] = round(info_D - info_D_values[col], 4)

for col, value in gain_values.items():
    print(f"Gain({col}): {value}")

max_gain_column = max(gain_values, key=gain_values.get)
max_gain_value = gain_values[max_gain_column]


print(f"\ninitial split is done for column: {max_gain_column} with Gain: {max_gain_value}")
```
